```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Đọc ảnh (giả sử tên file là 'dog.jpg')
img = cv2.imread('Picture1.jpg')

# 1. Kiểm tra chiều của ảnh
# Kết quả trả về là tuple: (Height, Width, Channels)
print("Kích thước ảnh (H, W, C):", img.shape)
```

```python
# Đọc ảnh
img = cv2.imread('Picture1.jpg')

if img is None:
    print("Không tìm thấy ảnh!")
else:
    # Lấy kích thước ảnh
    dimensions = img.shape
    height = img.shape[0]
    width = img.shape[1]
    channels = img.shape[2]

    print(f"Chiều của ảnh (Height, Width, Channels): {dimensions}")
    print(f"- Chiều cao: {height} pixel")
    print(f"- Chiều rộng: {width} pixel")
    print(f"- Số kênh màu: {channels}")
```

```python
# Cách hiển thị sai
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Ảnh gốc (BGR) - Bị sai màu")

# Cách hiển thị đúng:
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.subplot(1, 2, 2)
plt.imshow(img_rgb)
plt.title("Ảnh chuẩn (RGB) - Đúng màu")
plt.show()
```

```python
# Phóng to ảnh gấp 2 lần chiều ngang và dọc
img_zoomed = cv2.resize(img_rgb, None, fx=2.0, fy=2.0, interpolation=cv2.INTER_CUBIC)

print(f"Kích thước sau khi phóng to: {img_zoomed.shape}")

# Hiển thị
plt.imshow(img_zoomed)
plt.title("Ảnh đã phóng to 2x")
plt.show()
```

```python
# Lấy kích thước ảnh hiện tại
h, w, _ = img_rgb.shape

# Tính toán để cắt lấy vùng trung tâm
center_y, center_x = h // 2, w // 2
crop_h, crop_w = 300, 300  # Kích thước vùng muốn cắt

# Tính toạ độ cắt
y1 = center_y - crop_h // 2
y2 = center_y + crop_h // 2
x1 = center_x - crop_w // 2
x2 = center_x + crop_w // 2

# Thực hiện cắt
img_crop = img_rgb[y1:y2, x1:x2]

# Hiển thị
plt.imshow(img_crop)
plt.title("Ảnh đã crop")
plt.show()
```

```python
# Sử dụng ảnh đã crop ở Phần 4 để ghép (đảm bảo cùng kích thước)
# Ghép 2 ảnh theo chiều ngang
img_horizontal = cv2.hconcat([img_crop, img_crop])

# Ghép 2 ảnh theo chiều dọc
img_vertical = cv2.vconcat([img_crop, img_crop])

# Hiển thị kết quả
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_horizontal)
plt.title("Ghép ngang (Horizontal)")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_vertical)
plt.title("Ghép dọc (Vertical)")
plt.axis('off')

plt.show()
```
