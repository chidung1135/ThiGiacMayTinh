```python
import cv2
import matplotlib.pyplot as Dung
```

```python
img = cv2.imread('hinh2.jpg')
img = img[:,:,::-1]
Dung.imshow(img)
```

```python
img2 = cv2.blur(img, (11, 11))
Dung.imshow(img2)
```

```python
img3 = cv2.GaussianBlur(img, (11, 11),0)
Dung.imshow(img3)
```

```python
img4 = cv2.medianBlur(img, 15, 0)
Dung.imshow(img4)
```

```python
img_pepper01 = cv2.imread('pepper_noise01.jpg')
img_pepper01 = img_pepper01[:,:,::-1]
Dung.subplot(2,2,1)
Dung.imshow(img_pepper01)

img_pepper01a = cv2.blur(img_pepper01, (11, 11))
Dung.subplot(2,2,2)
Dung.imshow(img_pepper01a)

img_pepper01b = cv2.GaussianBlur(img_pepper01, (11, 11), 0)
Dung.subplot(2,2,3)
Dung.imshow(img_pepper01b)

img_pepper01c = cv2.medianBlur(img_pepper01, 15)
Dung.subplot(2,2,4)
Dung.imshow(img_pepper01c)
```
